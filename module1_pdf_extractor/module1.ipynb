```python
import fitz
import os
import json
```

```python
# Input PDF path
pdf_path = "Input/sample.pdf"

# Output folder for images
output_folder = "output_images"

# Create folder if not exists
os.makedirs(output_folder, exist_ok=True)

# Output JSON file
output_json = "module1_output.json"

print("Setup Completed Successfully!")
```

```python
# Open PDF
pdf_document = fitz.open(pdf_path)

# Count pages
total_pages = len(pdf_document)

print("Total Pages in PDF:", total_pages)
```

```python
module_output = []

# Process page by page
for page_index in range(len(pdf_document)):

    page = pdf_document[page_index]
    page_number = page_index + 1

    # Extract text
    extracted_text = page.get_text("text").strip()

    # Check images/diagrams in page
    images = page.get_images(full=True)

    # CASE 1: Page contains images/diagrams
    if len(images) > 0:

        image_filename = f"page_{page_number}.png"
        image_path = os.path.join(
            output_folder,
            image_filename
        )

        # Save page as image
        pix = page.get_pixmap()
        pix.save(image_path)

        page_data = {
            "page_number": page_number,
            "content_type": "Image",
            "extracted_content": extracted_text if extracted_text else None,
            "generated_image_path": image_path,
            "content_location": f"Page {page_number}"
        }

    # CASE 2: Text page
    else:

        page_data = {
            "page_number": page_number,
            "content_type": "Text",
            "extracted_content": extracted_text,
            "generated_image_path": None,
            "content_location": f"Page {page_number}"
        }

    module_output.append(page_data)

print("PDF Processing Completed Successfully!")
```

```python
with open(output_json, "w", encoding="utf-8") as file:
    json.dump(
        module_output,
        file,
        indent=4,
        ensure_ascii=False
    )

print("Module 1 Output Saved Successfully!")
```

```python
text_pages = 0
image_pages = 0

for page in module_output:
    if page["content_type"] == "Text":
        text_pages += 1
    else:
        image_pages += 1

print("Text Pages:", text_pages)
print("Image Pages:", image_pages)
```
